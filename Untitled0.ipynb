```python
import os
import base64
import openai
from gtts import gTTS
from IPython.display import Audio, display, Javascript
from google.colab import output # Apenas se usar no Google Colab

def configurar_ambiente():
    """Configura a chave de API da OpenAI de forma segura."""
    # Tenta buscar a chave nas variáveis do sistema
    api_key = os.getenv("OPENAI_API_KEY")

    if not api_key:
        # Se não houver chave no sistema, solicita ao usuário
        api_key = input("Por favor, insira sua OpenAI API Key: ")
        os.environ["OPENAI_API_KEY"] = api_key

    return openai.OpenAI(api_key=api_key)

def gravar_audio_colab(filename='input_audio.wav'):
    """Script em JavaScript para capturar áudio no ambiente do Google Colab."""
    print("Aguardando gravação...")
    js = Javascript('''
    async function recordAudio() {
        const div = document.createElement('div');
        const button = document.createElement('button');
        button.textContent = '🎤 CLIQUE PARA FALAR';
        button.style.cssText = 'padding: 15px 30px; background: #27ae60; color: white; border: none; border-radius: 8px; cursor: pointer; font-size: 16px; font-weight: bold;';
        document.body.appendChild(div);
        div.appendChild(button);

        const stream = await navigator.mediaDevices.getUserMedia({audio: true});
        const recorder = new MediaRecorder(stream);
        const chunks = [];

        recorder.ondataavailable = (e) => chunks.push(e.data);
        recorder.start();

        await new Promise(resolve => {
            button.onclick = () => {
                recorder.stop();
                button.textContent = '⌛ PROCESSANDO...';
                button.style.background = '#2980b9';
                button.disabled = true;
                resolve();
            };
        });

        const blob = new Blob(chunks);
        const reader = new FileReader();
        reader.readAsDataURL(blob);
        return new Promise(resolve => {
            reader.onloadend = () => resolve(reader.result);
        });
    }
    ''')
    display(js)
    data = output.eval_js('recordAudio()')
    binary = base64.b64decode(data.split(',')[1])
    with open(filename, 'wb') as f:
        f.write(binary)
    return filename

def processar_voz_ia(client, arquivo_audio):
    """Pipeline principal: Whisper -> GPT -> gTTS"""
    try:
        # 1. Whisper (Voz para Texto)
        print("1. Transcrevendo áudio com Whisper...")
        with open(arquivo_audio, "rb") as f:
            transcricao = client.audio.transcriptions.create(model="whisper-1", file=f)

        texto_usuario = transcricao.text
        print(f"🗣️ Você: {texto_usuario}")

        # 2. ChatGPT (Inteligência)
        print("2. Consultando ChatGPT...")
        resposta = client.chat.completions.create(
            model="gpt-3.5-turbo",
            messages=[{"role": "user", "content": texto_usuario}]
        )
        texto_ai = resposta.choices[0].message.content
        print(f"🤖 ChatGPT: {texto_ai}")

        # 3. gTTS (Texto para Voz)
        print("3. Gerando áudio de resposta...")
        tts = gTTS(text=texto_ai, lang='pt')
        tts.save("resposta.mp3")

        # Reproduzir no Colab
        display(Audio("resposta.mp3", autoplay=True))

    except Exception as e:
        print(f"❌ Ocorreu um erro técnico: {e}")

# --- EXECUÇÃO DO PROJETO ---
if __name__ == "__main__":1
    # Inicializa o cliente OpenAI
    client_openai = configurar_ambiente()

    # Executa o ciclo de voz
    audio_file = gravar_audio_colab()
    processar_voz_ia(client_openai, audio_file)
```

📖 Documentação do Projeto: Assistente de Voz Inteligente
1. Configuração do Ambiente e Bibliotecas
Nesta etapa inicial, preparamos o Python com as ferramentas necessárias. Instalamos bibliotecas para comunicação com a API da OpenAI, manipulação de áudio (pydub), conversão de texto em fala (gTTS) e integração com o Excel (openpyxl), caso seja necessário salvar logs de conversas futuramente.

2. Autenticação e Setup da API
Este bloco é responsável por estabelecer a conexão segura com os servidores da OpenAI. Utilizamos a biblioteca os para gerenciar a chave de API (API Key), que funciona como a identidade do desenvolvedor para acessar os modelos Whisper e GPT.

Código: Configuração do client = openai.OpenAI() e definição da OPENAI_API_KEY.

3. Captura de Áudio via Navegador (Interface JavaScript)
Como o Google Colab roda em servidores remotos, ele não acessa o microfone do seu computador diretamente. Por isso, utilizamos um script em JavaScript injetado no navegador.

Este código cria um botão dinâmico na tela.

Ao clicar, ele captura o fluxo de áudio (MediaRecorder), converte-o em um formato binário (Base64) e o envia de volta para o Python, salvando-o como um arquivo .wav.

4. Transcrição com OpenAI Whisper (Speech-to-Text)
O arquivo de áudio gravado é enviado para o modelo Whisper-1. O Whisper é uma rede neural de reconhecimento automático de fala (ASR) treinada em milhares de horas de áudio multilíngue. Sua função aqui é "ouvir" o áudio e transformá-lo em uma string de texto precisa, ignorando ruídos de fundo.

5. Processamento Inteligente com ChatGPT (LLM)
Com o texto da pergunta em mãos, o enviamos para o modelo GPT-3.5-Turbo.

Definimos um "System Prompt" para que a IA saiba que deve atuar como um assistente de voz conciso.

A IA analisa a pergunta e gera uma resposta textual coerente e inteligente.

6. Síntese de Voz com gTTS (Text-to-Speech)
A última etapa transforma a resposta textual da IA de volta em som. Utilizamos o gTTS (Google Text-to-Speech).

O texto é enviado ao motor de voz do Google.

Ele gera um arquivo .mp3 com entonação natural em português.

O comando Audio(autoplay=True) do IPython é executado para que você ouça a resposta imediatamente pelo alto-falante.
